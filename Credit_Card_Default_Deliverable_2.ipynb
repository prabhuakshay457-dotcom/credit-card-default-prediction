```python
# Install the Excel reader only if it is missing.
import importlib.util, subprocess, sys
if importlib.util.find_spec("xlrd") is None:
    subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "xlrd"])

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
import urllib.request, zipfile, io

source = "https://archive.ics.uci.edu/static/public/350/default+of+credit+card+clients.zip"
workbook = Path("default of credit card clients.xls")
if not workbook.exists():
    try:
        with urllib.request.urlopen(source, timeout=60) as response:
            archive_bytes = response.read()
        with zipfile.ZipFile(io.BytesIO(archive_bytes)) as archive:
            member = next(n for n in archive.namelist() if n.endswith(".xls"))
            workbook.write_bytes(archive.read(member))
    except Exception:
        print("Automatic download failed. Upload the extracted UCI .xls workbook.")
        from google.colab import files
        files.upload()
df = pd.read_excel(workbook, header=1)
print("Loaded:", df.shape)
```

## Data checks and feature selection
Planned lead: Akshay; peer review: Satya. Verify identifiers, repeated profiles, class balance, and monetary units. No undocumented outlier removal is performed.

```python
target = "default payment next month"
y = df[target]
print("Rows and columns:", df.shape)
print("Missing cells:", df.isna().sum().sum())
print("Duplicate IDs:", df.ID.duplicated().sum())
print("Repeated records excluding ID:", df.drop(columns="ID").duplicated().sum())
print(pd.DataFrame({
    "count": y.value_counts().sort_index(),
    "proportion": y.value_counts(normalize=True).sort_index()
}))
excluded = ["ID", target, "SEX", "EDUCATION", "MARRIAGE", "AGE"]
X = df.drop(columns=excluded)
cat = ["PAY_0", "PAY_2", "PAY_3", "PAY_4", "PAY_5", "PAY_6"]
num = [c for c in X if c not in cat]
print("Selected predictors:", len(X.columns))
print("Numeric:", num)
print("Categorical:", cat)
assert y.isin([0, 1]).all()
assert len(X.columns) == 19
```

## Training split and exploratory analysis
Planned EDA lead: Satya; peer review: Akshay. The 80/20 split is stratified. Explore training data only to avoid using the test partition to guide feature choices.

```python
from sklearn.model_selection import train_test_split
tr, te = train_test_split(
    np.arange(len(df)), test_size=0.2, stratify=y, random_state=42
)
print("Training:", len(tr), "Test:", len(te))
print("Training default rate:", y.iloc[tr].mean())
print("Training medians in New Taiwan dollars:")
print(df.iloc[tr][["LIMIT_BAL", "BILL_AMT1", "PAY_AMT1"]].median())
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
sns.countplot(x=y.iloc[tr], ax=axes[0], color="#527d9f")
axes[0].set(
    title="Training target distribution",
    xlabel="Next month default (0 No, 1 Yes)", ylabel="Clients"
)
rates = pd.DataFrame({
    "Status": df.iloc[tr].PAY_0, "Default": y.iloc[tr]
}).groupby("Status").Default.mean()
rates.plot.bar(ax=axes[1], color="#527d9f")
axes[1].set(
    title="Default rate by September status",
    xlabel="PAY_0 category", ylabel="Observed default rate"
)
fig.tight_layout()
plt.show()
```

## Preprocessing pipeline
Planned ownership: Akshay handles numeric scaling; Satya handles categorical encoding. Both review leakage prevention.

Keep the 35 repeated profiles because distinct IDs may represent different people; investigate profile grouping in Deliverable 2. Keep negative bill amounts because they may represent credits. Exclude demographic variables from prediction, retain them separately for later fairness auditing. Exclusion alone does not remove bias. Fit imputation, scaling, and encoding using training data only.


```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import make_pipeline
numeric = make_pipeline(
    SimpleImputer(strategy="median"), StandardScaler()
)
categorical = make_pipeline(
    SimpleImputer(strategy="most_frequent"),
    OneHotEncoder(handle_unknown="ignore")
)
prep = ColumnTransformer([
    ("numeric", numeric, num), ("status", categorical, cat)
])
print(prep)
```

## Initial model and baseline
Planned model lead: Akshay; metrics lead: Satya. Use default L2 regularization, C=1, no class weighting, and threshold 0.50. No model tuning has been performed.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.dummy import DummyClassifier
model = make_pipeline(
    prep, LogisticRegression(C=1.0, max_iter=2000, random_state=42)
)
model.fit(X.iloc[tr], y.iloc[tr])
baseline = DummyClassifier(strategy="most_frequent")
baseline.fit(X.iloc[tr], y.iloc[tr])
print("Training complete; all transformations fitted on training data.")
print("Classification threshold: 0.50")
```

```python
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, confusion_matrix
)
def evaluate(fitted):
    predictions = fitted.predict(X.iloc[te])
    probabilities = fitted.predict_proba(X.iloc[te])[:, 1]
    actual = y.iloc[te]
    return {
        "Accuracy": accuracy_score(actual, predictions),
        "Precision": precision_score(actual, predictions, zero_division=0),
        "Recall": recall_score(actual, predictions, zero_division=0),
        "F1": f1_score(actual, predictions, zero_division=0),
        "ROC AUC": roc_auc_score(actual, probabilities)
    }
results = pd.DataFrame({
    "Logistic regression": evaluate(model),
    "Majority baseline": evaluate(baseline)
}).T
print(results.to_string(float_format=lambda n: f"{n:.6f}"))
matrix = confusion_matrix(y.iloc[te], model.predict(X.iloc[te]))
print("Confusion matrix, rows actual and columns predicted:", matrix, sep="\n")
fig, ax = plt.subplots(figsize=(6, 4))
sns.heatmap(
    matrix, annot=True, fmt="d", cmap="Blues", cbar=False, ax=ax,
    xticklabels=["No default", "Default"],
    yticklabels=["No default", "Default"]
)
ax.set(
    title="Initial logistic regression test results",
    xlabel="Predicted label", ylabel="Actual label"
)
fig.tight_layout()
plt.show()
```

# Deliverable 2 Model refinement and evaluation
Training-only selection; historical test set reused from Deliverable 1.
Threshold constraints are illustrative and require stakeholder validation.

```python
from sklearn.base import clone
from sklearn.model_selection import GridSearchCV, StratifiedKFold
from sklearn.metrics import average_precision_score, fbeta_score, brier_score_loss
# Reserve a validation partition inside the original training partition.
fit_idx, val_idx = train_test_split(tr, test_size=0.2,
    stratify=y.iloc[tr], random_state=43)
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=44)
grid = {"logisticregression__C": [0.01, 0.1, 1.0, 10.0],
        "logisticregression__class_weight": [None, "balanced"]}
search = GridSearchCV(clone(model), grid, scoring="average_precision",
    cv=cv, n_jobs=1, refit=True, return_train_score=False)
search.fit(X.iloc[fit_idx], y.iloc[fit_idx])
cv_results = pd.DataFrame(search.cv_results_)
print("Fit/validation/test:",len(fit_idx),len(val_idx),len(te))
print("Selected parameters:",search.best_params_)
print(cv_results[["param_logisticregression__C",
    "param_logisticregression__class_weight","mean_test_score",
    "std_test_score"]].to_string(index=False))
```

```python
# Threshold selection uses validation data only.
# F2 emphasizes recall; illustrative constraints limit workload/false alerts.
pv = search.best_estimator_.predict_proba(X.iloc[val_idx])[:,1]
threshold_rows=[]
for threshold in np.arange(0.05,0.801,0.01):
    pred = (pv >= threshold).astype(int)
    threshold_rows.append({"threshold":float(threshold),
        "precision":precision_score(y.iloc[val_idx],pred,zero_division=0),
        "recall":recall_score(y.iloc[val_idx],pred),
        "F2":fbeta_score(y.iloc[val_idx],pred,beta=2),
        "flagged_fraction":float(pred.mean())})
threshold_table=pd.DataFrame(threshold_rows)
eligible=threshold_table.query("precision >= 0.40 and flagged_fraction <= 0.35")
assert len(eligible)>0, "No threshold meets the stated constraints."
chosen=eligible.sort_values(["F2","threshold"],ascending=[False,False]).iloc[0]
chosen_threshold=float(chosen.threshold)
print("Validation-selected operating point:")
print(chosen.to_string())
final_model=clone(search.best_estimator_).fit(X.iloc[tr],y.iloc[tr])
print("Final model refitted on",len(tr),"training rows.")
```

```python
# Final evaluation: no further model or threshold selection after this cell.
def score_prob(prob, threshold):
    pred=(prob>=threshold).astype(int)
    actual=y.iloc[te]
    return {"Accuracy":accuracy_score(actual,pred),
        "Precision":precision_score(actual,pred,zero_division=0),
        "Recall":recall_score(actual,pred),"F1":f1_score(actual,pred),
        "F2":fbeta_score(actual,pred,beta=2),
        "ROC_AUC":roc_auc_score(actual,prob),
        "Average_precision":average_precision_score(actual,prob),
        "Brier":brier_score_loss(actual,prob),
        "Flagged":int(pred.sum())}
pi=model.predict_proba(X.iloc[te])[:,1]
pf=final_model.predict_proba(X.iloc[te])[:,1]
comparison=pd.DataFrame({"Initial 0.50":score_prob(pi,0.50),
    "Tuned 0.50":score_prob(pf,0.50),
    "Final selected threshold":score_prob(pf,chosen_threshold)}).T
print(comparison.to_string(float_format=lambda v:f"{v:.4f}"))
final_pred=(pf>=chosen_threshold).astype(int)
final_cm=confusion_matrix(y.iloc[te],final_pred)
print("Final confusion matrix (rows actual, columns predicted):")
print(final_cm)
fig,axes=plt.subplots(1,2,figsize=(10,4))
for ax,prob,thr,title in zip(axes,[pi,pf],[0.5,chosen_threshold],
    ["Initial threshold 0.50",f"Final threshold {chosen_threshold:.2f}"]):
    sns.heatmap(confusion_matrix(y.iloc[te],prob>=thr),annot=True,
        fmt="d",cmap="Blues",cbar=False,ax=ax)
    ax.set(title=title,xlabel="Predicted default",ylabel="Actual default")
fig.tight_layout();plt.show()
```

```python
# Descriptive audit only: demographic variables were not model predictors.
audit=[]
for sex,label in [(1,"Male"),(2,"Female")]:
    mask=df.iloc[te].SEX.to_numpy()==sex
    actual=y.iloc[te].to_numpy()[mask];pred=final_pred[mask]
    tn,fp,fn,tp=confusion_matrix(actual,pred,labels=[0,1]).ravel()
    audit.append({"Group":label,"N":int(mask.sum()),
        "Defaults":int(actual.sum()),"Recall":tp/(tp+fn),
        "False_positive_rate":fp/(fp+tn),
        "Precision":tp/(tp+fp),"Flagged_rate":float(pred.mean())})
audit=pd.DataFrame(audit)
print(audit.to_string(index=False,float_format=lambda v:f"{v:.4f}"))
# Paired bootstrap uncertainty, conditional on this historical test sample.
rng=np.random.default_rng(45);actual=y.iloc[te].to_numpy()
deltas=[]
for _ in range(500):
    ix=rng.integers(0,len(te),len(te))
    deltas.append(recall_score(actual[ix],final_pred[ix])-
        recall_score(actual[ix],pi[ix]>=0.5))
print("Recall gain 95% percentile interval:",np.quantile(deltas,[0.025,0.975]))
print("Test set reused from Deliverable 1; external validation is still needed.")
```
